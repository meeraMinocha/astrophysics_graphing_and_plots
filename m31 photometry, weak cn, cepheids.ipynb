```python
from astropy.table import QTable
import matplotlib.pyplot as plt
import pandas as pd

m31_data = QTable.read("/Users/meera/Desktop/AST-04 Data/M31_Photometry_Complete.fits")
cn_data = QTable.read("/Users/meera/Desktop/AST-04 Data/M31_weak_cn.fits")
cepheid_data = pd.read_csv("/Users/meera/Desktop/AST-04 Data/M31_matches_after_2_offsets_07")
```

```python
m31_f475w = m31_data["f475w"] #blue
m31_f814w = m31_data["f814w"] #red

cn_f475w = cn_data["f475w"] #blue
cn_f814w = cn_data["f814w"] #red

cepheid_f475w = cepheid_data["F475W"]
cepheid_f814w = cepheid_data["F814W"]
```

```python
plt.figure(figsize = (8,6))
plt.scatter(m31_f475w-m31_f814w, m31_f814w, s=0.01, alpha = 0.1, c = "black")
plt.scatter(cn_f475w-cn_f814w, cn_f814w, s=2, c = "blue", label = "Weak CNs")
plt.scatter(cepheid_f475w-cepheid_f814w, cepheid_f814w, s=2, c = "red", label = "Cepheids")
plt.axhline(24, color = "red", linestyle = "--")
plt.xlim(-3,8)
plt.ylim(24, 17)
plt.legend(loc = "upper left")
plt.xlabel("F475W - F814W")
plt.ylabel("F814W")
plt.title("M31 color-magnitude diagram")
plt.grid(True)
plt.tight_layout()
plt.show()
```
