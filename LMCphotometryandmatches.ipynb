```python
from astropy.table import QTable
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
data = QTable.read("/Users/meera/Desktop/AST-04 Data/LMC_photometry.fit")
matches = pd.read_csv("/Users/meera/Desktop/AST-04 Data/LMC_matches_offset_updated.csv")
print(matches.columns)
```

```python
full_f475w = data["Bmag"].value
full_f814w = data["Imag"].value

matches_f475w = matches["Bmag"]
matches_f814w = matches["Imag"]

plt.figure(figsize = (8,6))
plt.scatter(full_f475w - full_f814w, full_f814w, s=0.01, c = "blue")
plt.scatter(matches_f475w - matches_f814w, matches_f814w, s=1, c = "red")
#plt.xlim(-2,8)
#plt.ylim(30, 20)
plt.gca().invert_yaxis()
plt.xlabel("F475W - F814W")
plt.ylabel("F814W")
plt.title("LMC Color-Magnitude Diagram")
plt.grid(True)
plt.tight_layout()
plt.show()
```
